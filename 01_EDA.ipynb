```python
# ============================================================
# 1. LOAD DATASET
# ============================================================

import pandas as pd
import numpy as np

# Menghubungkan Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Membaca dataset
df = pd.read_csv("/content/drive/MyDrive/Used_Car_Dataset.csv")

# Membuat salinan untuk proses cleaning
df_clean = df.copy()

# Menampilkan ukuran dataset
print("Ukuran dataset:", df.shape)

# Menampilkan 5 baris pertama
display(df.head())
```

```python
# ============================================================
# 2. INFORMASI DATASET
# ============================================================

# Menampilkan nama seluruh kolom
print("Nama kolom:")
print(df.columns.tolist())

# Menampilkan informasi dataset
print("\nInformasi dataset:")
df.info()

# Menampilkan statistik deskriptif
print("\nStatistik deskriptif:")
display(df.describe())
```

```python
# ============================================================
# 3. CEK MISSING VALUE
# ============================================================

# Menghitung jumlah missing value
missing = df.isna().sum().sort_values(ascending=False)

# Menampilkan hasil pemeriksaan
display(missing.to_frame("jumlah_missing"))

# Menampilkan total missing value
print("Total missing:", int(missing.sum()))
```

```python
# ============================================================
# 4. CEK DUPLIKAT
# ============================================================

# Menghitung jumlah data duplikat
jumlah_duplikat = df.duplicated().sum()

# Menampilkan hasil
print("Jumlah baris duplikat:", jumlah_duplikat)
```

```python
# ============================================================
# 5. DETEKSI OUTLIER DENGAN IQR
# ============================================================

# Menentukan kolom numerik yang diperiksa
outlier_columns = [
    "seats",
    "kms_driven",
    "mileage(kmpl)",
    "engine(cc)",
    "max_power(bhp)",
    "torque(Nm)",
    "price(in lakhs)"
]

# Membuat list untuk menyimpan hasil
hasil_outlier = []

# Melakukan deteksi outlier
for col in outlier_columns:

    # Menghitung kuartil pertama dan ketiga
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)

    # Menghitung IQR
    iqr = q3 - q1

    # Menentukan batas bawah dan atas
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    # Mengambil data kandidat outlier
    outlier = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    # Menyimpan hasil pemeriksaan
    hasil_outlier.append({
        "atribut": col,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "batas_bawah": lower,
        "batas_atas": upper,
        "jumlah_outlier": len(outlier)
    })

# Mengubah hasil menjadi DataFrame
outlier_result = pd.DataFrame(hasil_outlier)

# Menampilkan hasil
display(outlier_result)
```

```python
# ============================================================
# 6. CLEANING DATA
# ============================================================

# Menghapus kolom indeks yang tidak diperlukan
df_clean = df.drop(columns=["Unnamed: 0"], errors="ignore").copy()

# Menghapus data duplikat
df_clean = df_clean.drop_duplicates()

# Memisahkan kolom numerik dan kategorikal sementara
numeric_cols = df_clean.select_dtypes(include=np.number).columns
categorical_cols = df_clean.select_dtypes(include="object").columns

# Mengisi missing value numerik dengan median
for col in numeric_cols:
    median_value = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(median_value)

# Mengisi missing value kategorikal dengan modus
for col in categorical_cols:
    modus = df_clean[col].mode()[0]
    df_clean[col] = df_clean[col].fillna(modus)

# Memeriksa hasil cleaning
print("Missing value setelah cleaning:")
display(df_clean.isna().sum().to_frame("jumlah_missing"))

print("Jumlah data setelah cleaning:", df_clean.shape)
```

```python
# ============================================================
# 7. PEMISAHAN KOLOM NUMERIK DAN KATEGORIKAL
# ============================================================

# Mengambil kolom numerik
numeric_features = df_clean.select_dtypes(
    include=np.number
).columns.tolist()

# Mengambil kolom kategorikal
categorical_features = df_clean.select_dtypes(
    include="object"
).columns.tolist()

# Menampilkan hasil pemisahan
print("Kolom Numerik:")
print(numeric_features)

print("\nKolom Kategorikal:")
print(categorical_features)

print("\nJumlah kolom numerik:", len(numeric_features))
print("Jumlah kolom kategorikal:", len(categorical_features))
```
